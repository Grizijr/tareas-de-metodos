```python
import numpy as np
import matplotlib.pyplot as plt

#datos
m = 0.11       #masa
g = 9.8        #gravedad
k = 0.002      #K
v0 = 8.0       #velocidad inicial

# defino la ecuacion diferencial 
def f(t, v):
    return (-m*g - k*v*abs(v)) / m

#parametros metodo de euler
dt = 0.1                  #tiempo inicial
t_final = 1.0             #tiempo final 
n = int(t_final/dt) + 1   #numero de pasos
t = [i*dt for i in range(n)]
v = [0]*n

#C.I
v[0] = v0

#euler hacia adelante
for i in range(n-1):
    v[i+1] = v[i] + dt * f(t[i], v[i])

# Mostramos los resultados
print("Tiempo (s)    Velocidad (m/s)")
for i in range(n):
    print(f"{t[i]:5.1f}          {v[i]:.4f}")

#segundo punto que es encontrar la altura maxima
for i in range(1, n):
    if v[i] <= 0:
        print(f"\nEl proyectil alcanza su altura máxima aprox. en t = {t[i]:.1f} s")
        break
```
