```python
import numpy as np

def f(t, y):
    return (y/t) - (y/t)**2

def y_exacta(t):
    return t / (1 + np.log(t))

#datos
h = 0.1
t0, tf = 1, 2
t = np.arange(t0, tf + h, h)
n = len(t)

# Valores iniciales 
y = np.zeros(n)
y[0] = 1
y[1] = y[0] + h * f(t[0], y[0])  # primer valor con Euler

#predictor-corrector
def predictor_corrector(p):
    y_pc = y.copy()
    for i in range(1, n - 1):
        f_i = f(t[i], y_pc[i])
        f_i_1 = f(t[i-1], y_pc[i-1])
        
        #adams-bashforth 2 pasos
        y_pred = y_pc[i] + (h/2)*(3*f_i - f_i_1)
        
        #adams-moulton
        y_corr = y_pred
        for _ in range(p):
            f_pred = f(t[i+1], y_corr)
            y_corr = y_pc[i] + (h/12)*(5*f_pred + 8*f_i - f_i_1)
        
        y_pc[i+1] = y_corr
    return y_pc

#p= 2, 3, 4
for p in [2, 3, 4]:
    y_aprox = predictor_corrector(p)
    y_real = y_exacta(t)
    error = np.abs(y_real - y_aprox)
    print(f"\nResultados para p = {p}:")
    print(f"y(2) aproximado = {y_aprox[-1]:.6f},  y(2) exacto = {y_real[-1]:.6f}")
    print(f"Error global máximo = {np.max(error):.2e}")
```
