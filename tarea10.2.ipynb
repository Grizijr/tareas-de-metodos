```python
import numpy as np


#codigo para resolver una ecuacion de orden n

def rk4_n(f, a, b, y0, nsteps):
   
    h = (b - a) / nsteps
    t = np.linspace(a, b, nsteps + 1)
    n = len(y0)

    Y = np.zeros((nsteps + 1, n))
    Y[0, :] = y0

    def F(tt, yy):
        return np.concatenate((yy[1:], [f(tt, *yy)]))

    for k in range(nsteps):
        tk = t[k]
        yk = Y[k]

        k1 = F(tk, yk)
        k2 = F(tk + h/2, yk + (h/2) * k1)
        k3 = F(tk + h/2, yk + (h/2) * k2)
        k4 = F(tk + h,   yk + h * k3)

        Y[k+1] = yk + (h/6) * (k1 + 2*k2 + 2*k3 + k4)

    return t, Y

#ahora con el ejemplo dado  

def f(t, y, yp):
    return 2*yp - y + t*np.exp(t) - t
    
#condiciones iniciales

a = 0
b = 1
y0 = [0, 0]   
nsteps = 200

#resolver
t, Y = rk4_n(f, a, b, y0, nsteps)

#solucion analitica
y_exact = (1/6)*t**3 * np.exp(t) - t*np.exp(t) + 2*np.exp(t) - t - 2

#error
error = np.abs(Y[:,0] - y_exact)

#resultados
print("Valor aproximado en t = 1:", Y[-1,0])
print("Valor exacto en t = 1:", y_exact[-1])
print("Error:", error[-1])
```
